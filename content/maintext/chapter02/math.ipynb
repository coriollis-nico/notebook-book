Python incluye un módulo, `math`, con funciones matemáticas comunes
(las definidas en el estándar de C).
Este módulo debe estar disponible en cualquier instalación de Python.

````python
import math
````

````python
print("El número de Euler es", math.e)
print("La raszón entre el perímetro y el diámetro de un círculo es", math.pi)
````

````python
print("∛(27) =", math.cbrt(27))
print("e^1 =", math.exp(1))
print("2^10 =", math.exp2(10))
````

````python
print("sin(π/4) =", math.sin(math.pi / 4))
print("cos(π/4) =", math.cos(math.pi / 4))
print("tan(π/4) =", math.tan(math.pi / 4))
````

:::{dropdown} Funciones útiles de `math`

Por conveniencia listo aquí funciones útiles del módulo `math`.

| Nombre | Constante |
| --- | --- |
| `pi` | $\pi$ |

| Función | Resultado | Notas |

| `fsum(p)` | $\sum_i p_i$ | Suma números flotantes con mejor precisión que `sum` |

| `hypot(p)` | $\sqrt{\sum_i p_i^2}$ |
| `dist(p, q)` | $\sqrt{\sum_i (p_i - q_i)^2}$ |
| `isclose(a, b, [rel], [tol])` | $ a \approx b $ | `True` o `False`, útil para comparar `float`s |
