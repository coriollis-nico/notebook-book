# Ejemplo: resorte de Hooke (oscilador armónico simple)

> Una masa de $450\ \text{g}$ está sujeta a un resorte de rigidez (_i.e._ de constante hookeana)
> $ k = 500\ \text{N}/\text{m} $.
> Cuando empiezas a observar esta masa ($t = 0$) esta se encuentra a $-35\ \text{cm}$
> de su punto de equilibrio y se mueve con velocidad de
> de $ + 4\ \text{m}/\text{s} $.
>
> Calcula su posición $1\ \text{s}$ después, $5\ \text{s}$ después, y
> $10\ \text{s}$ después.

La ecuación diferencial para resortes simples (_i.e._ de un oscilador armónico) es
$$
  \ddot{x}(t) = -\frac{k}{m} x(t)
$$
que tiene solución general
$$
  x(t) = A \sin{(\omega t)} + B \cos{(\omega t)}. \tag{$\omega = \sqrt{k/m}$}
$$
Las condiciones de frontera $ x(0) = x_i $ y $ \dot{x}(0) = v_i $ nos permiten reescribir
la solución como
$$
  x(t) = \frac{v_i}{\sqrt{k/m}} \sin{(\sqrt{k/m}\ t)} + x_i \cos{(\sqrt{k/m}\ t)}.
$$

Para calcular las soluciones numéricas, traduzcamos nuestro resultado general a una
función de Python.

```python
import math
```

```python
def eom_general(m, xi, vi, k, t):
    """
    Ecuación de movimiento general para oscilador armónico.

    Parámetros
    ----------
    m   : masa [kg]
    xi  : posición inicial [m]
    vi  : velocidad inicial [m/s]
    k   : cte. de rigidez del resorte [N/m]
    t   : tiempo [s]

    Regresa
    -------
    x   : posición [m] al tiempo t
    """

    omega = math.sqrt(k / m)

    x = (vi / (omega) * math.sin(omega * t)) + (xi * math.cos(omega * t))

    return x
```

Por conveniencia, como las condiciones del problema están dadas, podemos crear
una segunda función con estas condiciones ya definidas.

```python
m = 0.450  # kg
k = 500  # N/m
xi = -0.35  # m
vi = 4  # m/s
t_p = [1, 5, 10]  # s
```

```python
x = lambda t: eom_general(m, xi, vi, k, t)
```

y ahora sí, calculemos las soluciones:

```python
xs = [x(t) for t in t_p]
```

```python
print(xs)

print("Con formato y con el tiempo correspondiente:")

for i in range(len(t_p)):
    print(f"x({t_p[i]}s) = {xs[i]:.2f} m")
```
