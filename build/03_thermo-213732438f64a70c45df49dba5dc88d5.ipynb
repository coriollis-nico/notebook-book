> Calcula el trabajo requerido para comprimir 2 mol de un gas ideal que permanece a una
> temperatura constante de 20°C desde un volumen de 4 litros a uno final de 1 litro.
>
> -- Zemansky & Dittman, p. 57

Al ser un gas ideal,
$$ p V = n R T. $$
El trabajo está dado por
$$ W = - \int_{V_i}^{V_f} p \, \mathrm{d}V. $$
Con una sustitución simple,
$$
  W = - n R T \ln{\left( \frac{V_f}{V_i} \right)}
$$

```python
import math
```

```python
R: float = 8.315  # J/mol/K
```

```python
n: float = 2
Vi: float = 0.001 * 4.0
# 4 L en m^3
Vf: float = 0.001
# 1 L en m^3
T: float = 20 + 273.15
```

```python
def work_qsit(vi: float, vf: float, n: float, T: float) -> float:
    """
    Trabajo en expansión/compresión isotérmica cuasi-estática.

    Calcula el trabajo realizado/expandido al expandir/comprimir n moles de un gas ideal
    a temperatura constante T.
    """

    work: float = -n * R * T * math.log(vf / vi)
    return work
```

```python
print(f"El trabajo requerido para comprimir el gas es {work_qsit(Vi, Vf, n, T):.1f} J.")
```
