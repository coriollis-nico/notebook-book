### NumPy

Implementemos una distribución gausiana no normalizada:
$$ f(x) = \exp{(-x^2)} $$

```python
def normal(x):
  "Distribución gaussiana (no normalizada)"
  return np.exp(-np.square(x))
```

Implementemos una función sigmoide
$$
\begin{align*}
  f(x) &= \frac{1}{1 + \mathrm{e}^{-x}} \\
    &= \frac{1}{1 + \exp{(-x)}}
\end{align*}
$$

```python
def sigmoid(x):
  "Función sigmoide simple"
  return 1./(1. + np.exp(-x))
```

y aplicamos estas funciones a un arreglo de 1000 puntos entre -10 y 10.

```python
X = np.linspace(-10, 10, 1000)
Y1 = normal(X)
Y2 = sigmoid(X)
```

### Matplotlib

Graficamos `normal(X)` y `sigmoid(X)`.

```python
fig, ax = plt.subplots(layout="constrained")

ax.plot(X, Y1, label="Gausiana")
ax.plot(X, Y2, label="Sigmoide")

ax.set_xlabel("x")
ax.set_ylabel("y(x)")

ax.set_xlim(X[0], X[-1])

ax.legend(loc="best")
```
